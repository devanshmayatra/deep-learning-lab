# PReLU: Forward Computation and Manual Gradients

## Objective
Implement PReLU derivatives, distinguish input gradients from parameter gradients, and verify the results against autograd.

## Activation

$$
f(x)=
\begin{cases}
x & x>0\\
ax & x\leq0
\end{cases}
$$

For upstream gradient \(g\), away from zero:

$$
\frac{\partial L}{\partial x}
=
\begin{cases}
g & x>0\\
ag & x<0
\end{cases}
$$

For a shared scalar \(a\):

$$
\frac{\partial L}{\partial a}
= \sum_{i:x_i<0} g_i x_i.
$$

## Test Case
- Inputs: `[-3, -1, 2, 4]`
- Negative slope: `a = 0.2`
- Upstream gradients: `[4, -2, 3, 1]`

## Verified Results
- Forward outputs: `[-0.6, -0.2, 2, 4]`
- Input gradients: `[0.8, -0.4, 3, 1]`
- Shared slope gradient: `-10`

Both manual gradient comparisons against autograd returned `True`.

## Key Distinction
Input gradients remain separate for each input. Contributions to the same shared parameter are summed; input and parameter gradients are not added together.

## Scope
The practical required guidance on gradient separation and shared-parameter accumulation. The autograd comparison code was supplied during review. The test excludes zero inputs.

## Reference
He et al. — *Delving Deep into Rectifiers*.

```python
import torch
```

```python
x        = [-3, -1, 2, 4]
a        = 0.2
upstream = [ 4, -2, 3, 1]
```

```python
input_grads = []
a_contribution = 0.0

for i in range(len(x)):
    x_e = x[i]
    grad = upstream[i]
    
    if x_e > 0:
        input_grads.append(grad)
    else:
        dlda = x_e * grad
        dldx = a * grad
        input_grads.append(dldx)
        a_contribution += dlda

a_contribution, input_grads
```

```python
x_auto = torch.tensor(x, dtype=torch.float64, requires_grad=True)
a_auto = torch.tensor(a, dtype=torch.float64, requires_grad=True)
upstream_auto = torch.tensor(upstream, dtype=torch.float64)

y = torch.where(x_auto > 0, x_auto, a_auto * x_auto)
y.backward(upstream_auto)

print("Forward:", y.detach())
print("Autograd input gradients:", x_auto.grad)
print("Autograd a gradient:", a_auto.grad)

print("Input gradients match:", torch.allclose(
    torch.tensor(input_grads, dtype=torch.float64),
    x_auto.grad
))
print("a gradient matches:", torch.allclose(
    torch.tensor(a_contribution, dtype=torch.float64),
    a_auto.grad
))
```
