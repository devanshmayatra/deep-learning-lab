```python
import torch
```

```python
x        = [-3, -1, 2, 4]
a        = 0.2
upstream = [ 4, -2, 3, 1]
```

```python
input_grads = []
a_contribution = 0.0

for i in range(len(x)):
    x_e = x[i]
    grad = upstream[i]
    
    if x_e > 0:
        input_grads.append(grad)
    else:
        dlda = x_e * grad
        dldx = a * grad
        input_grads.append(dldx)
        a_contribution += dlda

a_contribution, input_grads
```
