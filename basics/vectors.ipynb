```python
import numpy as np
import torch
```

```python
batch_size = 32
sequence_length = 5
vocab_size = 20
embedding_dim = 6
```

```python
embedding_table = torch.randn([vocab_size,embedding_dim])
token_table = torch.randint(0,20,[batch_size, sequence_length])

embedding_table.shape, token_table.shape
```

```python
lookup_table = embedding_table[token_table]

lookup_table.shape
```

```python
one_hot = torch.nn.functional.one_hot(
    token_table, num_classes=vocab_size
).to(dtype=embedding_table.dtype)

one_hot_embeddings = one_hot @ embedding_table

print(one_hot.shape)             # (32, 5, 20)
print(one_hot_embeddings.shape)  # (32, 5, 6)
print(torch.allclose(lookup_table, one_hot_embeddings))
```

```python
weights = torch.randn([6,8])
bias = torch.zeros([8,])
```

```python
output = lookup_table @ weights + bias
```

```python
output.shape
```

```python
avg = output.mean(1, keepdim=False)
flatten = output.reshape(output.shape[0], -1)

avg.shape, flatten.shape
```

```python
batch_size = 3
sequence_length = 4
vocab_size = 9
embedding_dim = 5
output_features = 2
```

```python
embedding_table = torch.randn([vocab_size,embedding_dim])
token_table = torch.randint(0,vocab_size,[batch_size, sequence_length])

embedding_table.shape, token_table.shape
```

```python
lookup_table = embedding_table[token_table]

lookup_table.shape
```

```python
one_hot = torch.nn.functional.one_hot(
    token_table, num_classes=vocab_size
).to(dtype=embedding_table.dtype)

one_hot_embeddings = one_hot @ embedding_table

print(one_hot.shape)
print(one_hot_embeddings.shape)
print(torch.allclose(lookup_table, one_hot_embeddings))
```

```python
weights = torch.randn([lookup_table.shape[-1],output_features])
bias = torch.zeros([output_features])

weights.shape
```

```python
output = lookup_table@weights + bias

output.shape
```
