```python
import torch
```

```python
fan_in = 100
fan_out = 100
batch_size = 256

base_scale = 0.01
xavier_init = 0.1
standard = 1.0

g = torch.Generator().manual_seed(42)

input = torch.randn([256, 100], generator=g) 
W1 = torch.randn([100, 100], generator=g)
W2 = torch.randn([100, 100], generator=g)
W3 = torch.randn([100, 100], generator=g)
W4 = torch.randn([100, 100], generator=g)
W5 = torch.randn([100, 100], generator=g)

layers = [input, W1, W2, W3, W4, W5]
```

```python
base_scaled_layers = [l * base_scale for l in layers[1:]]
```

```python
def calc_mean_std(layers):
    prev_output = input

    for i, weight in enumerate(layers, start=1):
        output = torch.tanh(prev_output @ weight)
        prev_output = output

        mean = output.mean()
        std = output.std()
        saturation = (output.abs() > 0.99).float().mean()

        print(
            f"Layer {i} | mean={mean.item():.6f} | "
            f"std={std.item():.6f} | saturation={saturation.item():.4%}"
        )
```

```python
calc_mean_std(base_scaled_layers)
```

```python
xavier_scaled_layers = [l * xavier_init for l in layers[1:]]
```

```python
calc_mean_std(xavier_scaled_layers)
```

```python
standard_scaled_layers = [l * standard for l in layers[1:]]
```

```python
calc_mean_std(standard_scaled_layers)
```
