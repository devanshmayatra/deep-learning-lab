# MiniNN: CIFAR-10 Training Example

This notebook demonstrates training a small residual CNN using my
custom MiniNN library, built on PyTorch tensors and autograd.

## Workflow

1. Create the training, validation, and test loaders.
2. Build the model with an explicit device and random generator.
3. Create Adam once before training.
4. Train with augmentation and evaluate on validation data.
5. Save the best model and optimizer state.
6. Restore the checkpoint for inference or continued training.

## Baseline result

- Fresh training run: 10 epochs
- Optimizer: Adam, learning rate 0.001
- Training augmentation: random horizontal flips
- Best checkpoint: epoch 10
- Validation accuracy: 64.54%
- Validation loss: 1.0259
- Test set: not evaluated for this run

## Checkpoint notes

Recreate the same architecture before loading.
